```python
marks=[20,40,80,50,90]
marks.sort()
print("Ascending order:",marks)
print("Lowest marks:",marks[0])
print("Highest marks:",marks[4])
for i in marks:
    print(i)
total=0
for num in marks:
    total+=num
print("Total:",total)
avr=total/len(marks)
print("Average :",avr)
```
