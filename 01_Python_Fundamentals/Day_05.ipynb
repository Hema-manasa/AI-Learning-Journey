```python
def find_maximum(a,b,c):
    if a>b and a>c:
        return a
    elif b>c and b>a:
        return b
    else:
        return c
print(find_maximum(10,20,30)) 
find_maximum(20,90,40)
```

```python
def calculate_average(marks):
    tot=sum(marks)
    average=tot/len(marks)
    return average
def grade(average):

    if average >= 90:
        return "A Grade"

    elif average >= 75:
        return "B Grade"

    else:
        return "C Grade"
def marksheet(student):
    average=calculate_average(student["marks"])
    print("Name:",student["name"])
    print("Average marks:",average)
    print("Grade:",grade(average))
marksheet(student={"name":"hema","marks":[20,30,40]})    
```
